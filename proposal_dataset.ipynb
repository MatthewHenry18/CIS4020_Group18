Loading and basic overview of the dataset we chose.

**Data:** download data and point `DATA_DIR` at the CSVs
(`https://archive.ics.uci.edu/dataset/891/cdc+diabetes+health+indicators`).

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", 40)
print("pandas", pd.__version__)
```

```python
# link to download dataset:
# https://archive.ics.uci.edu/dataset/891/cdc+diabetes+health+indicators
DATA_DIR = "./data/"    # put data set in this folder

FILES = {"diabetes_012":"diabetes_012_health_indicators_BRFSS2015.csv",
         "diabetes_binary_5050split":"diabetes_binary_5050split_health_indicators_BRFSS2015.csv",
         "diabetes_binary":"diabetes_binary_health_indicators_BRFSS2015.csv"}

def load_oulad():
    missing = [f for f in FILES.values() if not os.path.exists(os.path.join(DATA_DIR, f))]
    if missing:
        raise FileNotFoundError(
            f"Couldn't find these OULAD files in DATA_DIR={DATA_DIR!r}: {missing}. "
            "Set DATA_DIR to the folder containing the CSVs."
        )
    print("[info] loading real OULAD from", DATA_DIR)
    return {k: pd.read_csv(os.path.join(DATA_DIR, f)) for k, f in FILES.items()}

tables = load_oulad()
for name, t in tables.items():
    print(f"{name:22s} {t.shape}")
```

```python
for name, df in tables.items():
    print("-------------------------------------------------")
    print(f"table name: {name}")
    print("shape (rows, columns):", df.shape)
    print("\ncolumn types:")
    print(df.dtypes)
    print("\nfirst rows:")
    display(df.head())
```
